## Holiday (holiday)

```python
print("Unique values are: ")
print(df_raw['holiday'].unique())

print("-" * 50)

print("Unique value count is: ")
print(df_raw['holiday'].nunique())

print("-" * 50)

print("Duplicate row count: ")
print(df_raw['holiday'].duplicated().sum())
```

```python
# Calculate counts and percentages
holiday_count = df_raw['holiday'].value_counts().reindex([0, 1], fill_value=0)
holiday_percentage = holiday_count / holiday_count.sum() * 100

# Readable category labels
labels = ['Non-Holiday', 'Holiday']
values = holiday_count.values
colors = ['#36B9A8', '#FF6B6B']

# Create graph
plt.figure(figsize=(10, 6))

bars = plt.bar(
    labels,
    values,
    color=colors,
    width=0.55,
    edgecolor='white',
    linewidth=2
)

# Add record count and percentage labels
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f'{values[i]:,}\n({holiday_percentage.iloc[i]:.1f}%)',
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

# Graph labels and styling
plt.xlabel('Day Type', fontsize=12, fontweight='bold', labelpad=10)
plt.ylabel('Number of Records', fontsize=12, fontweight='bold', labelpad=10)

plt.title(
    'Holiday vs Non-Holiday Distribution',
    fontsize=16,
    fontweight='bold',
    pad=20
)

plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.gca().set_axisbelow(True)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

The time-series investigation identified 165 missing timestamps. These missing hours are a sequence-continuity concern because they may interrupt consecutive hourly observations. ### 4. Target Leakage
The `casual` and `registered` features must be excluded when forecasting total demand (`cnt`) in advance if they represent rental counts observed during the target hour. In the standard dataset, total demand is the sum of these two components. Including them would leak information about the target into the model inputs.

The random-split Random Forest experiment achieved MAE = 43.16 and RMSE = 66.05, compared with MAE = 140.88 and RMSE = 179.36 for the mean baseline. These results are exploratory and are not the official forecasting benchmark because a random split may not adequately reflect prediction on future observations.

### 6. Preprocessing and Modeling Decisions
Before modeling, validate hour values, investigate missing timestamps, preserve chronological order, and define a leakage-safe feature set. The official benchmark should use a chronological train/test split and a suitable time-aware baseline. Feature availability at prediction time and the effect of incomplete hourly sequences must also be considered.

## `Step 3 (C):Generate sequences only when time continuity is valid`

```python
feature_cols = [
    "cnt",          # Historical demand only
    "temp",
    "atemp",
    "hum",
    "windspeed",
    "hr",
    "weekday",
    "mnth",
    "season",
    "workingday",
    "holiday",
    "weathersit",
]
```

```python
def create_hourly_sequences(
        df,
        feature_col,
        target_col='cnt',
        window_size=24
):
    df = df.sort_values("timestamp").reset_index(drop=True).copy()

    if df["timestamp"].duplicated().any():
        raise ValueError("Duplicate timestamps detected.")

    X_sequences = []
    y_values = []
    target_times = []

    one_hour = pd.Timedelta(hours=1)

    for tartget_idx in range(window_size, len(df)):
        input_start = tartget_idx - window_size
        input_end = tartget_idx - 1

        input_times = df['timestamp'].iloc[input_start:tartget_idx]
        input_differences = input_times.diff().dropna()

        if not (input_differences == one_hour).all():
            continue

        target_time = df['timestamp'].iloc[tartget_idx]
        last_input_time = df['timestamp'].iloc[input_end]

        if target_time - last_input_time != one_hour:
            continue

        X_window = df[feature_col].iloc[input_start:tartget_idx]

        if X_window.isna().any().any():
            continue

        target_value = df[target_col].iloc[tartget_idx]

        if pd.isna(target_value):
            continue

        X_sequences.append(X_window.to_numpy(dtype = np.float32))
        y_values.append(float(target_value))
        target_times.append(target_time)

    X = np.asarray(X_sequences, dtype=np.float32)
    y = np.asarray(y_values, dtype=np.float32)
    target_times = pd.DatetimeIndex(target_times)

    return X, y, target_times




```

```python
X, y, target_times = create_hourly_sequences(
    df=df_eval,
    feature_col=feature_cols,
    target_col='cnt',
    window_size=24
)

print("Input shape:", X.shape)
print("Target shape:", y.shape)
print("Valid training examples:", len(y))
```

```python

import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Timestamp-indexed observed rental counts
demand_by_time = (
    df_eval
    .set_index("timestamp")["cnt"]
    .sort_index()
)

if demand_by_time.index.has_duplicates:
    raise ValueError("Duplicate timestamps must be resolved first.")

# Use the existing validation target timestamps
validation_target_times = pd.DatetimeIndex(val_times)

# Same hour on the previous day
yesterday_times = (
    validation_target_times - pd.Timedelta(hours=24)
)

# Retrieve actual demand at target time and 24 hours earlier
actual_counts = demand_by_time.reindex(validation_target_times)
yesterday_counts = demand_by_time.reindex(yesterday_times)

# A valid prediction requires both observations
yesterday_valid = (
    actual_counts.notna().to_numpy()
    & yesterday_counts.notna().to_numpy()
)

y_yesterday_actual = actual_counts.to_numpy()[yesterday_valid]
y_yesterday_pred = yesterday_counts.to_numpy()[yesterday_valid]
yesterday_valid_times = validation_target_times[yesterday_valid]

if len(y_yesterday_actual) == 0:
    raise ValueError("No eligible validation pairs for yesterday baseline.")

# Metrics for the same-hour-yesterday baseline
yesterday_mae = mean_absolute_error(
    y_yesterday_actual,
    y_yesterday_pred
)

yesterday_rmse = np.sqrt(
    mean_squared_error(
        y_yesterday_actual,
        y_yesterday_pred
    )
)

print("Eligible yesterday-baseline predictions:", len(y_yesterday_actual))
print(f"Same-hour-yesterday MAE:  {yesterday_mae:.2f}")
print(f"Same-hour-yesterday RMSE: {yesterday_rmse:.2f}")
```

```python

# Eligible timestamps for the previous-hour baseline
previous_hour_times = (
    validation_target_times - pd.Timedelta(hours=1)
)

previous_counts = demand_by_time.reindex(previous_hour_times)
actual_counts = demand_by_time.reindex(validation_target_times)

previous_valid = (
    actual_counts.notna().to_numpy()
    & previous_counts.notna().to_numpy()
)

previous_valid_times = validation_target_times[previous_valid]

# Intersection: only timestamps eligible for BOTH baselines
common_times = previous_valid_times.intersection(
    yesterday_valid_times
).sort_values()

common_actual = demand_by_time.reindex(common_times).to_numpy()

previous_predictions = demand_by_time.reindex(
    common_times - pd.Timedelta(hours=1)
).to_numpy()

yesterday_predictions = demand_by_time.reindex(
    common_times - pd.Timedelta(hours=24)
).to_numpy()

assert (
    np.isfinite(common_actual).all()
    and np.isfinite(previous_predictions).all()
    and np.isfinite(yesterday_predictions).all()
)

results = pd.DataFrame([
    {
        "Model": "Previous-hour persistence",
        "Valid predictions": len(common_times),
        "MAE": mean_absolute_error(common_actual, previous_predictions),
        "RMSE": np.sqrt(mean_squared_error(common_actual, previous_predictions)),
    },
    {
        "Model": "Same-hour-yesterday",
        "Valid predictions": len(common_times),
        "MAE": mean_absolute_error(common_actual, yesterday_predictions),
        "RMSE": np.sqrt(mean_squared_error(common_actual, yesterday_predictions)),
    },
])

print(results.to_string(index=False))
```

```python

# Regularization strengths to experiment with
alphas = [0.01, 0.1, 1, 10, 100, 1000]

ridge_results = []
ridge_models = {}

for alpha in alphas:

    # Create model
    model = Ridge(alpha=alpha)

    # Fit using training data ONLY
    model.fit(X_train_flat, y_train)

    # Predict validation demand
    val_predictions = model.predict(X_val_flat)

    # Calculate validation metrics
    mae = mean_absolute_error(y_val, val_predictions)

    rmse = np.sqrt(
        mean_squared_error(y_val, val_predictions)
    )

    mse = mean_squared_error(y_val, val_predictions)

    # Store results
    ridge_models[alpha] = model

    ridge_results.append({
        "Alpha": alpha,
        "MAE": mae,
        "RMSE": rmse,
        "MSE": mse
    })

# Display comparison
ridge_results_df = pd.DataFrame(ridge_results)

ridge_results_df = ridge_results_df.sort_values(
    "MAE"
).reset_index(drop=True)

print(ridge_results_df.to_string(index=False))
```
